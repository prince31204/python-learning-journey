## Sets

#### A sets is an unordered collection of items. Every set element is uniques (no duplicates) and must be immutable cannot be chnaged however a set iteslef is mutable we can add or remove items from it sets can also be used to perform mathematical set operations like union, intersection,symmetric differences, etc
##### Characterstics
##### unordered,mutable, no duplicates, cant't contsin mutable data types

```python
# empty
s = set()
print(type(s))
# 1D And 2D 
s1 = {1,2,3,4}
print(s1)
#2 D
#Homo
s3 = {1,'hello',4.5, True}
print(s3)
# Using type conversion
s4 = set([1,2,3])
print(type(s4))
# Duplicates not allowed
s5 = {1,2,1,2,3,3}
print(s5)
# sets cant have mutable items
```

```python
s1 = {1,2,3}
s2 = {3,2,1}
print(s1 == s2)
```

### Acessing items

```python
s = {1,2,3,4}
s.add(5)
print(s)
#update
s.update([6,7,8])
print(s)
```

```python
# del
s = {1,2,3,4}
print(s)
```

```python
# discard
s.discard(5)
print(s)
```

### Sets Operations

```python
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
# Union(|)

# Intersection(&)
s1 & s2
# Difference
s1 - s2
s2-s1
# Symmetric Difference
s1 ^ s2

```

### Sets Function

```python
s = {3,1,4,5,2,7}
len(s)
sum(s)
```

```python
# Union /update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
s1.union(s1)
s1.update(s2)
print(s1)
print(s2)

```

```python
# intersection / intersection-update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
s1.intersection(s2)
s1.intersection_update(s2)
print(s1)
print(s2)
```

```python
# difference / differnce -update
# intersection / intersection-update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
s1.difference(s2)
s1.difference_update(s2)
print(s1)
print(s2)
```

```python
# symmetric difference
# intersection / intersection-update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}
s1.symmetric_difference(s2)
s1.symmetric_difference_update(s2)
print(s1)
print(s2)
```

```python
# isdisjoint/issubset/issuperset
s1 = {1,2,3,4}
s2= {7,8,5,6}
s1.isdisjoint(s2)


```

```python
#subset
s1 ={1,2,3,4,5}
s2 = {3,4,5}
s2.issubset(s1)
```
