## Membership Operators

```python
# Arithmetic operator
print(5+6)
print(5-6)
print(5*6)
print(5/2)
print(5//2)
print(5%2)
print(5**2)
```

```python
# Relational Operator
print(4>5)
print(4<5)
print(4>=4)
print(4<=4)
print(4==4)
print(4!=4)
```

```python
# Logical Operator
print(1 and 0)
print(1 or 0)
print(not 0)
```

```python
# Bitwise Operator
print(2 & 3)
# Bitwise or
print (2|3)
#Bitwise xor
print(2^3)
# Bitwise not
print(~3)
```

```python
# Assignment Operators
a=2
# print(a)

a = 2
# a = a%2
a %=2

#a++ ++a this notation not used in python
a =a +2
print(a)
```

### Membership operator

```python
# in/not in
print('D' not in 'Delhi')
print(1 in [2,3,4,5,6]) 
```
