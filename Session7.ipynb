## File handling serialization and deserialization

#### Some theory

### Type of data used for I/O
#### text -1,2,3,4,5 as a sequence of unicode chars
#### Binary -1,2,3,4,5 as a sequence of bytes of its binary equivalent


#### Hence there are 2 type files to deal with 
### Text files- All programs files are text files
#### Binary files- images, music,videos,exe files

### How file I/O  is done in most programming languages
### Open a file
### Read / Write a data
### Close the file

### Writing t a file

```python
# Case1 - If the file is not present
f= open('sample.txt','w')
f.write('Hello world')
f.close()
```

```python
# Write multiple strings
f= open('sample1.txt','w')
f.write('hello world')
f.write('\nhow are you')
f.close()
```

```python
# Case -2 If the file is already present
f = open('sample.txt','w')
f.write('prince parmar')
f.close()
```

```python
# Problem with w mode
# introducing append mode
f = open('sample.txt','a')
f.write('\nI AM FINE')
f.close()
```

```python
# Write lines
l = ['hello\n', 'hi\n ']
f= open('sample.txt','w')
f.writelines(l)
f.close()
```

```python
# reading from files
f = open('sample.txt','r')
s= f.read()
print(s)
f.close()
 
```

```python
# readline()-> to read line by line
f = open('sample.txt','r')
print(f.readline(),end='')
print(f.readline(),end='')
f.close()

```
